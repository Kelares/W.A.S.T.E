```python
import os, shutil
import splitfolders

# 1️⃣ Create a combined folder in working (read-write)
combined = "/kaggle/working/all_classes"
os.makedirs(combined, exist_ok=True)

# 2️⃣ Copy your garbage classes in
shutil.copytree(
    "/kaggle/input/garbage-classification-v2/garbage-dataset",
    os.path.join(combined),
    dirs_exist_ok=True
)

# 3️⃣ Copy your background images in as their own class
shutil.copytree(
    "/kaggle/input/background/background",
    os.path.join(combined, "background"),
    dirs_exist_ok=True
)

shutil.copytree(
    "/kaggle/input/hgi-30-classification/classification/carton",
    os.path.join(combined, "cardboard"),
    dirs_exist_ok=True   # merge into existing, instead of erroring
)

shutil.copytree(
    "/kaggle/input/hgi-30-classification/classification/glassbottle",
    os.path.join(combined, "glass"),
    dirs_exist_ok=True   # merge into existing, instead of erroring
)

shutil.copytree(
    "/kaggle/input/hgi-30-classification/classification/glassbottle",
    os.path.join(combined, "glass"),
    dirs_exist_ok=True   # merge into existing, instead of erroring
)

shutil.copytree(
    "/kaggle/input/hgi-30-classification/classification/plasticbottle",
    os.path.join(combined, "plastic"),
    dirs_exist_ok=True   # merge into existing, instead of erroring
)

shutil.copytree(
    "/kaggle/input/hgi-30-classification/classification/sodacan",
    os.path.join(combined, "metal"),
    dirs_exist_ok=True   # merge into existing, instead of erroring
)

shutil.copytree(
    "/kaggle/input/hgi-30-classification/classification/tincan",
    os.path.join(combined, "metal"),
    dirs_exist_ok=True   # merge into existing, instead of erroring
)

shutil.copytree(
    "/kaggle/input/hgi-30-classification/classification/wasterpaper",
    os.path.join(combined, "paper"),
    dirs_exist_ok=True   # merge into existing, instead of erroring
)


# 4️⃣ Now split into train/val 80/20
splitfolders.ratio(
    input=combined, 
    output="garbage_dataset-split", 
    seed=42, 
    ratio=(.8, .2)
)
```

```python
from ultralytics import YOLO

model = YOLO("yolo11s-cls.pt") 

model.train(
    data='garbage_dataset-split',
    imgsz=224,
    batch=32,
    device=[0,1],
    epochs=300,
    patience=5,
    optimizer='AdamW',

    # Colour
    hsv_h=0.05, hsv_s=0.7, hsv_v=0.5,

    # Geometry
    degrees=180, translate=0.10, scale=0.50,
    fliplr=0.5, flipud=0.25, shear = 5.0, perspective = 0.0005,

    # Occlusion & policy
    erasing=0.4,
)
```
